```python
from torchvision.io.image import read_image
from torchvision.models.segmentation import fcn_resnet50, FCN_ResNet50_Weights
from torchvision.transforms.functional import to_pil_image
from PIL import ImageDraw

import torch
import numpy as np

# Load the image
img = read_image("Datasets/FSC147_384_V2/images_384_VarV2/7519.jpg")

# Initialize the model with the best available weights
weights = FCN_ResNet50_Weights.DEFAULT
model = fcn_resnet50(weights=weights)
model.eval()

# Apply inference preprocessing transforms
preprocess = weights.transforms()
batch = preprocess(img).unsqueeze(0)

# Perform prediction using the model
with torch.no_grad():
    prediction = model(batch)['out'][0]

# Softmax to get probabilities
normalized_masks = prediction.softmax(dim=0)

# Get category indices and names
class_to_idx = {cls: idx for idx, cls in enumerate(weights.meta["categories"])}

# Convert original image to PIL for drawing
original = to_pil_image(img)
draw = ImageDraw.Draw(original)

# Loop over each class, extract bounding boxes, and draw them
for cls, idx in class_to_idx.items():
    mask = normalized_masks[idx].numpy()
    mask = (mask > 0.5)  # Threshold to create a binary mask
    pos = np.where(mask)

    if pos[0].size > 0 and pos[1].size > 0:
        xmin, xmax = np.min(pos[1]), np.max(pos[1])
        ymin, ymax = np.min(pos[0]), np.max(pos[0])
        draw.rectangle([xmin, ymin, xmax, ymax], outline="red", width=2)

# Show the resulting image with bounding boxes
original.show()

# Optionally, print categories
print(weights.meta["categories"])
```

```python
from torchvision.io.image import read_image
from torchvision.models.segmentation import fcn_resnet50, FCN_ResNet50_Weights
from torchvision.transforms.functional import to_pil_image
from PIL import Image, ImageDraw, ImageColor

import torch
import numpy as np
import matplotlib.pyplot as plt

# Load the image
img = read_image("Datasets/FSC147_384_V2/images_384_VarV2/549.jpg")

# Initialize the model with the best available weights
weights = FCN_ResNet50_Weights.DEFAULT
model = fcn_resnet50(weights=weights)
model.eval()

# Apply inference preprocessing transforms
preprocess = weights.transforms()
batch = preprocess(img).unsqueeze(0)

# Perform prediction using the model
with torch.no_grad():
    prediction = model(batch)['out'][0]

# Softmax to get probabilities
normalized_masks = prediction.softmax(dim=0)

# Get category indices and names
class_to_idx = {cls: idx for idx, cls in enumerate(weights.meta["categories"])}

if 'car' in class_to_idx:
    car_index = class_to_idx['car']
    car_mask = normalized_masks[car_index].numpy()

    # Applying a more dynamic threshold might help highlight cars better
    threshold = car_mask.mean() + car_mask.std()  # Mean + 1 Standard Deviation Threshold
    car_mask_thresholded = car_mask > threshold

    # Convert the boolean mask to uint8
    car_mask_uint8 = (car_mask_thresholded * 255).astype(np.uint8)  # Convert boolean mask to uint8

    # Create a PIL image from the uint8 mask
    car_mask_image = to_pil_image(car_mask_uint8)  # Create an image from the uint8 mask
    car_mask_image = car_mask_image.convert("RGBA")
    color = ImageColor.getrgb("blue")  # Color for the car mask
    overlay = Image.new("RGBA", car_mask_image.size, color + (0,))
    draw = ImageDraw.Draw(overlay)
    draw.bitmap((0, 0), car_mask_image, fill=color + (128,))  # Semi-transparent blue overlay

    # Convert original image to PIL for blending
    original = to_pil_image(img)

    # Resize overlay to match original image and blend
    overlay = overlay.resize(original.size)
    original.paste(overlay, (0, 0), overlay)

# Show the resulting image with the 'car' mask
original.show()

# Optionally, print categories
print(weights.meta["categories"])


```

```python
import torch
from torchvision.models.detection import fasterrcnn_resnet50_fpn, FasterRCNN_ResNet50_FPN_Weights
from torchvision.io import read_image
from torchvision.transforms.functional import convert_image_dtype

# Load the image
img = read_image("Datasets/FSC147_384_V2/images_384_VarV2/7182.jpg")

# Initialize the model with the best available weights
weights = FasterRCNN_ResNet50_FPN_Weights.DEFAULT
model = fasterrcnn_resnet50_fpn(weights=weights)
model.eval()

# Preprocess the image (Faster R-CNN models expect float images in [0, 1])
img = convert_image_dtype(img, dtype=torch.float)

# Perform inference
with torch.no_grad():
    prediction = model([img])

# Filter results for 'kite' and count them
kite_count = 0
for element in prediction[0]['labels']:
    if weights.meta['categories'][element] == 'car':
        kite_count += 1

print(f"Detected {kite_count} kites in the image.")
weights.meta['categories']
```

```python
# Example of sliding window
import torch
from torchvision.transforms import functional as F

def sliding_window(image, step=64, window_size=(256, 256)):
    """ Slide a window across the image """
    for y in range(0, image.shape[1], step):
        for x in range(0, image.shape[2], step):
            new_img = image[:, y:y+window_size[1], x:x+window_size[0]]
            if new_img.shape[1] == window_size[1] and new_img.shape[2] == window_size[0]:
                yield new_img, x, y

# Dummy image tensor
image = torch.randn(3, 800, 1200)  # Simulate an image

# Process each window
for window, x, y in sliding_window(image):
    window = F.resize(window, size=(224, 224))  # Resize to fit ResNet50 input
    output = resnet50(window.unsqueeze(0))  # Classify with ResNet50
```
