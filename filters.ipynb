```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
input_image_path = "2022-07-14 14.39.14.jpeg"
img = cv2.imread(input_image_path, cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
plt.axis('off')  # Optional: to not show axes for a cleaner image
plt.show()
```

```python
# Gaussian kernel size (must be an odd number)
kernel_size = 5

# Create the Gaussian kernel
kernel = np.ones((kernel_size, kernel_size), np.float32) / (kernel_size ** 2)
```

```python
Identity_kernel = np.array([
    [0, 0, 0],
    [0, 1, 0],
    [0, 0, 0]
], dtype=np.float32)

Identity_move_left_kernel = np.array([
    [0, 0, 0],
    [1, 0, 0],
    [0, 0, 0]
], dtype=np.float32)

horizontal_edge_kernel = np.array([
    [-1, -2, -1],
    [0, 0, 0],
    [1, 1, 1]
], dtype=np.float32)

vertical_kernel = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=np.float32)

reduce_brightness_kernel = np.array([
    [0, 0, 0],
    [0, 0.5, 0],
    [0, 0, 0]
], dtype=np.float32)

compute_horzontal_kernel = np.array([
    [0, 0, 0],
    [-1, 0, 1],
    [0, 0, 0]
], dtype=np.float32)

compute_horzontal_kernel = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
], dtype=np.float32)
```

```python
identity_image = cv2.filter2D(img, -1, Identity_kernel)
identity_left_image = cv2.filter2D(img, -1, Identity_move_left_kernel)
horizontal_edge_image = cv2.filter2D(img, -1, horizontal_edge_kernel)
reduce_brightness_image = cv2.filter2D(img, -1, reduce_brightness_kernel)
computer_horizontal_image = cv2.filter2D(img, -1, compute_horzontal_kernel)
```

```python

# Assume img1, img2, img3, img4 are your images

plt.figure(figsize=(10, 10))  # Optional: Set the figure size


plt.subplot(2, 3, 1)  # Second subplot in a 2x2 grid
plt.imshow(identity_image, cmap='gray')
plt.title('Identity Original')
plt.axis('off')

plt.subplot(2, 3, 2)  # Second subplot in a 2x2 grid
plt.imshow(identity_left_image, cmap='gray')
plt.title('Move Pixel to Right')
plt.axis('off')

plt.subplot(2, 3, 3)  # Third subplot in a 2x2 grid
plt.imshow(horizontal_edge_image, cmap='gray')
plt.title('Horizontal Edge')
plt.axis('off')

plt.subplot(2, 3, 4)  # Fourth subplot in a 2x2 grid
plt.imshow(reduce_brightness_image, cmap='gray')
plt.title('Reduce Brightness 0.5')
plt.axis('off')


plt.subplot(2, 3, 5)  # Fourth subplot in a 2x2 grid
plt.imshow(computer_horizontal_image, cmap='gray')
plt.title('Compute Horizontal')
plt.axis('off')

plt.tight_layout()  # Optional: Improve layout
plt.show()
```
